# Data Exploration

Initial notebook for exploring the incident report dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ==========================================================
# PROJECT PATHS
# ==========================================================

# In Jupyter notebooks, __file__ is not defined.
# Use __file__ when available, otherwise fall back to the current working directory.
if "__file__" in globals():
    project_root = Path(__file__).resolve().parent.parent
else:
    project_root = Path.cwd().resolve()

    # Walk upward until we find the project root that contains data/raw
    for candidate in [project_root, *project_root.parents]:
        if (candidate / "data" / "raw").exists():
            project_root = candidate
            break

input_file = (
    project_root
    / "data"
    / "raw"
    / "Police_Department_Incident_Reports__2018_to_Present_20260725.csv"
)

output_dir = project_root / "outputs"
output_dir.mkdir(parents=True, exist_ok=True)

# ==========================================================
# LOAD DATASET
# ==========================================================

print("=" * 60)
print("LOADING DATASET")
print("=" * 60)

df = pd.read_csv(input_file)

# ==========================================================
# DATASET SHAPE
# ==========================================================

print("\nDataset Shape")
print("-" * 60)

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

# ==========================================================
# FIRST 10 RECORDS
# ==========================================================

print("\nFirst 10 Records")
print("-" * 60)

print(df.head(10))

# ==========================================================
# DATASET INFO
# ==========================================================

print("\nDataset Information")
print("-" * 60)

df.info()

# ==========================================================
# MISSING VALUES
# ==========================================================

print("\nMissing Values")
print("-" * 60)

missing = df.isnull().sum()

print(missing)

# ==========================================================
# DUPLICATE RECORDS
# ==========================================================

print("\nDuplicate Records")
print("-" * 60)

duplicates = df.duplicated().sum()

print(f"Total Duplicate Rows : {duplicates}")

# ==========================================================
# DESCRIPTIVE STATISTICS
# ==========================================================

print("\nDataset Statistics")
print("-" * 60)

print(df.describe(include="all"))

# ==========================================================
# INCIDENT CATEGORY DISTRIBUTION
# ==========================================================

print("\nTop Incident Categories")
print("-" * 60)

category_counts = df["Incident Category"].value_counts()

print(category_counts)

# ==========================================================
# POLICE DISTRICT DISTRIBUTION
# ==========================================================

print("\nPolice District Distribution")
print("-" * 60)

district_counts = df["Police District"].value_counts()

print(district_counts)

# ==========================================================
# SAVE SUMMARY
# ==========================================================

summary_file = output_dir / "data_summary.txt"

with open(summary_file, "w", encoding="utf-8") as f:

    f.write("DATA EXPLORATION REPORT\n")
    f.write("=" * 60 + "\n\n")

    f.write(f"Rows    : {df.shape[0]}\n")
    f.write(f"Columns : {df.shape[1]}\n\n")

    f.write("Missing Values\n")
    f.write("-" * 60 + "\n")
    f.write(missing.to_string())
    f.write("\n\n")

    f.write("Duplicate Records\n")
    f.write("-" * 60 + "\n")
    f.write(f"{duplicates}\n\n")

    f.write("Top Incident Categories\n")
    f.write("-" * 60 + "\n")
    f.write(category_counts.to_string())
    f.write("\n\n")

    f.write("Police District Distribution\n")
    f.write("-" * 60 + "\n")
    f.write(district_counts.to_string())

print("\nSummary Saved")

print(summary_file)

# ==========================================================
# CATEGORY DISTRIBUTION CHART
# ==========================================================

top_categories = category_counts.head(10)

plt.figure(figsize=(12,6))

plt.bar(top_categories.index, top_categories.values)

plt.xticks(rotation=45, ha="right")

plt.title("Top 10 Incident Categories")

plt.xlabel("Incident Category")

plt.ylabel("Number of Incidents")

plt.tight_layout()

category_chart = output_dir / "top10_categories.png"

plt.savefig(category_chart)

plt.close()

print("Top Category Chart Saved")

# ==========================================================
# DISTRICT DISTRIBUTION CHART
# ==========================================================

top_districts = district_counts.head(10)

plt.figure(figsize=(12,6))

plt.bar(top_districts.index, top_districts.values)

plt.xticks(rotation=45, ha="right")

plt.title("Police District Distribution")

plt.xlabel("Police District")

plt.ylabel("Number of Incidents")

plt.tight_layout()

district_chart = output_dir / "district_distribution.png"

plt.savefig(district_chart)

plt.close()

print("District Chart Saved")

# ==========================================================
# FINISHED
# ==========================================================

print("\n" + "=" * 60)
print("DATA EXPLORATION COMPLETED")
print("=" * 60)

print("\nGenerated Files")

print("✔", summary_file)

print("✔", category_chart)

print("✔", district_chart)

print("\nDone Successfully!")
```
